Update old ConSinGAN requirements

```python
from pathlib import Path
import re

p = Path("requirements.txt")
text = p.read_text()

# Convert package==old.version into package>=old.version
text = re.sub(r"^([A-Za-z0-9_.-]+)==([^\n]+)$", r"\1>=\2", text, flags=re.MULTILINE)

p.write_text(text)
```
